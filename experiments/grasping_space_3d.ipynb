## Grasping Configuration checker

Hand frame must be located at a certain height range from the cap (and rotation, so a transform), fingers at an offset (also a range) and allow full rotation of gripper hand and cap

```python
# Get panda_hand frame between 0.0105 m and 0.11 m above the cap base_link frame
# Fingers should be open between 0.024 and 0.025 m and -0.024 and -0.025 m
# Get a function that checks whether these constraints are satisfied

E = plant.GetFrameByName("base_wsg", gripper)  # End-effector frame
Cap = plant.GetFrameByName("base_link", cap)    # Cap frame

def check_grasp_constraints(q):
    
    plant.SetPositions(plant_context, q)
    diagram.ForcedPublish(diagram_context)
    # Get the current pose of the hand frame
    X_WE = plant.CalcRelativeTransform(plant_context, plant.world_frame(), E)
    X_WCap = plant.CalcRelativeTransform(plant_context, plant.world_frame(), Cap)
    
    # Compute the relative transform from Cap to E
    X_CapE = X_WCap.inverse().multiply(X_WE)
    
    z_height = X_CapE.translation()[2]
    
    # right_finger_pos = right_finger_joint.get_translation(plant_context)
    left_finger_pos  = left_finger_joint.get_translation(plant_context)
    
    # Check height constraint
    height_ok = 0.06 <= z_height <= 0.07
    
    # Check finger opening constraints
    fingers_ok = (-0.024 <= left_finger_pos <= -0.025)
    
    
    return height_ok and fingers_ok, z_height, left_finger_pos
```

```python
q_grasp = np.array([0, 0, -0.024])  # Example configuration

print("Checking grasp constraints for q_grasp:", check_grasp_constraints(q_grasp))
```

```python
# Define the 8 corner points of the convex hull
x_bounds = [-3.14, 3.14]
y_bounds = [-1, 1]
z_bounds = [-0.055, -0.024]

lower_joint_limits = np.array([x_bounds[0], y_bounds[0], z_bounds[0]])
upper_joint_limits = np.array([x_bounds[1], y_bounds[1], z_bounds[1]])

z_bounds_grasp = [-0.025, -0.024]

# Generate all corner points
placement_points = np.array([[x, y, z] for x in x_bounds for y in y_bounds for z in z_bounds])
grasp_points = np.array([[x, y, z] for x in x_bounds for y in y_bounds for z in z_bounds_grasp])
# placement_points = np.array([[x, y] for x in x_bounds for y in z_bounds])
# grasp_points = np.array([[x, y] for x in x_bounds for y in z_bounds_grasp])

# Compute the convex hull
placement_hull = ConvexHull(placement_points)
grasp_hull = ConvexHull(grasp_points)

# Convert ConvexHull to HPolyhedron
def convex_hull_to_hpolyhedron(hull):
    A = hull.equations[:, :-1]
    b = -hull.equations[:, -1]
    return HPolyhedron(A, b)

placement_polytope = convex_hull_to_hpolyhedron(placement_hull)
grasp_polytope = convex_hull_to_hpolyhedron(grasp_hull)
```

```python
builder = visualizer.builder
plant = visualizer.plant
scene_graph = visualizer.scene_graph
q_star = visualizer.q_star
rat_fk = visualizer.rat_forward_kin
inspector = visualizer.model_inspector
diagram = visualizer.task_space_diagram
context = visualizer.task_space_diagram_context
cspace_free_polytope = visualizer.cspace_free_polytope
```

```python

%tb
# Grasp Constraints

# --- Build the IK program over q (this is what IRIS-ZO will check) ---
ik_ctx = plant.CreateDefaultContext()       # separate context for thread-safety
ik = InverseKinematics(plant, ik_ctx)
q = ik.q()

# Frames you already have:
E   = plant.GetFrameByName("base_wsg", gripper)  # hand frame
Cap = plant.GetFrameByName("base_link", cap)     # cap frame

# 1) Position of the hand origin in the cap frame: keep z in [0.06, 0.07] m,
#    leave x,y mostly free (use a reasonable box).
xy_span = 0.15  # tighten later if you want the hand centered
ik.AddPositionConstraint(
    frameB=E, p_BQ=np.zeros(3),                 # point Q = origin of E
    frameA=Cap,
    p_AQ_lower=np.array([-xy_span, -xy_span, 0.060]),
    p_AQ_upper=np.array([ +xy_span, +xy_span, 0.070]),
)

# 2) Orientation: keep the hand’s axes aligned with the cap within a small angle.
#    (This constrains the relative rotation between Cap and E to be within theta_bound.)
ik.AddOrientationConstraint(
    frameAbar=Cap,                R_AbarA=RotationMatrix(),     # desired = Cap orientation
    frameBbar=E,                  R_BbarB=RotationMatrix(),     # E identity
    theta_bound=np.deg2rad(5)     # <= 5 degrees misalignment
)

# 3) Left finger joint window (right finger welded => no constraint for it).
left_finger_joint = plant.GetJointByName("left_finger_sliding_joint", gripper)
iL = left_finger_joint.position_start()
# Match your predicate: left finger in [0.024, 0.025] m
ik.prog().AddBoundingBoxConstraint(
    np.array([0.024]), np.array([0.025]), np.array([q[iL]])
)



# IRIS-ZO Setup

generator = RandomGenerator(1234)
checker = SceneGraphCollisionChecker(
    model=diagram,
    robot_model_instances=[plant.GetModelInstanceByName("robot"), plant.GetModelInstanceByName("bottle_cap")],
    edge_step_size=0.01)

common_sampled_iris_options = CommonSampledIrisOptions()
common_sampled_iris_options.prog_with_additional_constraints = ik.prog()

iris_zo_options = IrisZoOptions()
iris_zo_options.sampled_iris_options = common_sampled_iris_options


clique_cover_options = IrisFromCliqueCoverOptions()
clique_cover_options.coverage_termination_threshold = 0.9
clique_cover_options.iris_options = iris_zo_options
clique_cover_options.num_points_per_visibility_round = 200

grasp_regions = IrisInConfigurationSpaceFromCliqueCover(
    checker=checker,
    options=clique_cover_options,
    generator=generator,
    sets=[]    
)
```
